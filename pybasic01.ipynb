## Python入门练习(输入、输出、字符串、for循环)

 ### 1.计算1-100所有整数和，并打印结果

```python
sum = 0
# range()函数生成一个整数序列，从0开始，到指定的结束值（不包含结束值）,内置函数
for i in range(1, 101):
    sum += i
print(sum)
```

### 2.统计1-4组成的互不相同且无重复的三位数

```python
count = 0
for i in range(1, 5):
    for j in range(1, 5):
        for k in range(1, 5):
            if i != j and i != k and j != k:
                print(i, j, k)
                count += 1
# 打印统计结果
print(f"1-4组成的互不相同且无重复的三位数有{count}个")
```

### 3.打印九九乘法表

```python
for i in range(1, 10):
    for j in range(1,10):
        if(j<=i):
            print(f"{i}*{j}={i*j}\t",end="")
    print("\n")
```

### 4.找出并打印所有的水仙花数

```python
# 一个三位数，其个位数字立方和等于该数本身
for i in range(1,10):
    for j in range(1,10):
        for k in range(1,10):
            if((i**3+j**3+k**3) == int(f"{i}{j}{k}")):
                print(int(f"{i}{j}{k}"))
```

### 5.猴子吃桃，猴子从第一天每天吃总桃数的一半多一个，第十天只剩一个，求第一天桃子数

```python
sum = 1
for i in range(9,0,-1):
    sum+=1
    sum*=2
print(f"第一天桃子的总数是{sum}")
```

### 6.输入一行字符，分别统计出其中英文字母、空格、数字和其他字符的个数。

```python
s = input("请输入一个字符串：")
letter = 0
space = 0
digit = 0
other = 0
for c in s:
    if c.isalpha():
        # print("英文字母：",c)
        letter+=1
    elif c.isspace():
        # print("space:",c)
        space+=1
    elif c.isdigit():
        # print("digit:",c)
        digit+=1
    else:
        # print("other:",c)
        other+=1
print(s)
print(f"英文字母有{letter}个，空格字符有{space}个，数字有{digit}个，其他有{other}个")
```
