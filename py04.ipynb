### Python入门(时间)

#### 1.时间戳

```python
import time
current_time = time.time() #获取当前时间戳，自纪元以来的秒数(1970-01-01 00:00:00)
# print(current_time)
current_ctime = time.ctime() #获取当前时间戳的本地时间字符串
# print(current_ctime)
specified_current_ctime = time.ctime(1717102524) #获取指定时间戳的本地时间字符串
# print(specified_current_ctime)
current_local_struct_time = time.localtime() #可传入参数，默认当前时间戳的struct_time对象(年月日时分秒)
# print(current_local_struct_time)
current_utc_time = time.gmtime() #可传参，获取世界标准UTC当前时间戳的struct_time对象
# print(current_utc_time)
current_format_time = time.asctime(time.localtime()) #将struct_time时间戳对象转换成指定格式的字符串表示(格式为：星期 月份 日 时:分:秒 年)
print(current_format_time)
```

#### 2.计算耗时时间

```python
import time
# start = time.time()
start = time.perf_counter() #返回一个高精度、单调递增的时钟值
for i in range(3000):
    print(i)
# end = time.time()
end = time.perf_counter()
print(f"遍历从0-2999消耗的总时间为：{end-start:.3f}秒")
```

#### 3.延时操作

```python
import time
print(time.strftime("%Y-%m-%d %H:%M:%S",time.localtime()))
time.sleep(1) #将程序暂停指定秒数
print(time.strftime("%Y-%m-%d %H:%M:%S",time.localtime()))
```

#### 4.实现简单秒表

```python
import time
print("按下回车开始计时，中断程序停止")
input("准备中...")
start = time.time()
print("开始计时...")
try:
    while True:
        cross_time = time.time()-start
        print(f"\r计时：{cross_time:.0f}秒",end="") #\r从开头覆盖上一次输出
        time.sleep(1)
except KeyboardInterrupt:
    #捕捉用户结束的信号
    end = time.time()
    total_time = round(end-start,2) #round函数四舍五入保留两位小数
    print(f"\n计时结束，运行总时间为：{total_time}秒")
```
