#### 1.翻转头尾并拼接字符串

```python
def reverse_and_contact(s,d):
    if d > len(s):
        raise Exception("输入数字大于字符串长度！") # 抛出自定义异常 Exception是异常类型
    header_filpped = s[:d][::-1]
    header_remaining = s[d:]
    header_result = header_filpped + header_remaining
    tail_filpped = s[-d:][::-1]
    tail_remaining = s[:-d]
    tail_result = tail_filpped + tail_remaining
    #print("头部翻转拼接的结果为：",header_result)
    #print("尾部翻转拼接的结果为：",tail_result)
    return header_result,tail_result
s = input("请输入一串字符串：")
d = int(input("请输入要翻转的字符串长度："))
try:
    print(reverse_and_contact(s,d))
except Exception as e:
    print(e)
```

#### 2.绘制螺旋图案

```python
from tkinter import*     # 导入tkinter模块，用于创建图形用户界面
canvas = Canvas(width=500,height=400) #设置宽高
canvas.pack(expand=True,fill=BOTH) #pack布局管理将Canvas组件添加到窗口当中，设置其扩展和填充属性

k = 1 #控制圆圈的半径
j = 1 #控制k的增量
for i in range(26):
    canvas.create_oval(250-k,200-k,250+k,200+k,width=1) #绘制圆圈
    k+=j
    j+=0.3
mainloop()
```

#### 3.随机生成

```python
import random
number = random.random() #随机生成0.0-1.0（不包括）的小数
print(number)
number1 = random.randint(0,10) #随机生成0-10的整数
print(number1)
list1 = [1,2,3,4,5]
random_el = random.choice(list1) #从给定序列中随机选择一个元素
print(random_el)
list2 = ["a","b","c","d","e","f"]
random.shuffle(list2) #将给定序列(只能列表)随机排序，无返回值
print(list2)
```

#### 4.随机生成验证码

```python
import random
import string #引入模块
# 定义验证码中可能包含的字符类型
character = string.ascii_letters + string.digits
l = int(input("请输入你想要的验证码长度："))
verify_code = ""
for i in range(l):
    verify_code += random.choice(character) 
print(f"生成的验证码：{verify_code}")
```
