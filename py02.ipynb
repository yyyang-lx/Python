## python入门（列表、字典排序、函数）

#### 1.学生成绩排序

```python
student = [
    {'sno':101,'sname':"z",'sgrade':88},
    {'sno':102,'sname':"q",'sgrade':66},
    {'sno':103,'sname':"s",'sgrade':77},
    {'sno':104,'sname':"l",'sgrade':99},
]
student_sort = sorted(student,key=lambda x: x["sgrade"],reverse=True) #reverse默认为false升序
print(f"原列表为：{student}")
print(f"排序后列表为：{student_sort}")
```

#### 2.字典排序

```python
people = {'张三':18,'李四':30,'王五':24}
max_age = float('-inf') # 初始年龄最大值为负无穷
max_name = ""
for name,age in people.items(): #对字典进行解包,items()表示的是元组
    if age>max_age:
        max_age = age
        max_name = name
print(f"年龄最大者的姓名：{max_name}，年龄：{max_age}")
```

#### 3.按序插入列表

```python
li = [22,33,44,55,66,78]
num = int(input("请输入一个数："))
for i in range(len(li)):
    if li[i] > num:
        li.insert(i,num)
        break
else: #不能写在循环里面
    li.append(num)
print(li)
```

#### 4.计算数字列表总和

```python
def sum_of_list(numbers):
    sum = 0
    for number in numbers:
        sum+=number
    return sum
list1 = [1,2,3,4,5]
list2 = [3,1,4,6,7]
print(sum_of_list(list1))
print(sum_of_list(list2))
print(sum(list1))
print(sum(list2))
```

#### 5.移除列表中的多个元素

```python
def remove_el(list1,list2):
    for i in list1:
        if i in list2:
            list1.remove(i)
    return list1
list1 = [3,5,7,9,11,13]
list2 = [7,11]
# print(f"从列表{list1}中移除列表{list2}，结果为：{remove_el(list1,list2)}")
# 用列表推导式
print(f"从列表{list1}中移除列表{list2}，结果为：{[i for i in list1 if i not in list2]}")
```

#### 6.列表元素去重

```python
def remove_du(list1):
    result = []
    for i in list1:
        if i not in result:
            result.append(i)
    return result
list1 = [10,30,20,10,40,40,80]
# print(f"对列表去重后的结果：{remove_du(list1)}")
# 快速去重==>set 集合支持自动去重操作
print(list(set(list1)))
```

#### 7.计算字典中所有值的总和

```python
def sum_of_dict(dict1):
    # 获取字典中所有值
    sum = 0
    for value in dict1.values():
        sum+=value
    return sum
    
example_dict = {"a":100,"b":200,"c":300}
print(f"该字典所有值的和为：{sum_of_dict(example_dict)}")
```

#### 8.找到两个数组中的公共子前缀

```python
def find_comment(A,B):
    n = len(A)
    ans = [] # 列表储存最后结果子前缀个数
    cntA = set()# 调用set函数，创建一个空集合，集合的表示方式：{}
    cntB = set()
    cnt = set()
    for i in range(n): # 只用传下标
        cntA.add(A[i]) # 集合中添加元素(不可重复)
        cntB.add(B[i])
        if B[i] in cntA:
            cnt.add(B[i])
        if A[i] in cntB:
            cnt.add(A[i]) # 存储公共子前缀
        ans.append(len(cnt)) # 记录前i+1个元素公共子前缀的个数
    return ans
A = input("请输入列表A:").split()
B = input("请输入列表b:").split()
print(f"列表的公共子前缀为：{find_comment(A,B)}")
```
